# Customer Retention & Churn Analysis

## Future Interns – Data Science & Analytics Task 2

This project analyzes customer churn, retention patterns, customer lifetime, and factors associated with customer churn using Python and Power BI.

```python
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df["TotalCharges"].isna().sum()
```

```python
df[df["TotalCharges"].isna()][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
df = df.dropna(subset=["TotalCharges"])
```

```python
df.info()
```

```python
df["Churn"].value_counts()
```

```python
churn_rate = (df["Churn"] == "Yes").mean() * 100

churn_rate
```

```python
df["Contract"].value_counts()
```

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn
```

```python
df["tenure"].describe()
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
)
```

```python
df.groupby("TenureGroup", observed=True)["Churn"].value_counts(normalize=True) * 100
```

```python
df["MonthlyCharges"].describe()
```

```python
df["ChargeGroup"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0, 40, 80, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
df.groupby("ChargeGroup", observed=True)["Churn"].value_counts(normalize=True) * 100
```

```python
df["InternetService"].value_counts()
```

```python
df.groupby("InternetService", observed=True)["Churn"].value_counts(normalize=True) * 100
```

```python
df["TechSupport"].value_counts()
```

```python
df.groupby("TechSupport", observed=True)["Churn"].value_counts(normalize=True) * 100
```

```python
df["OnlineSecurity"].value_counts()
```

```python
df.groupby("OnlineSecurity", observed=True)["Churn"].value_counts(normalize=True) * 100
```

```python
df["PaymentMethod"].value_counts()
```

```python
df.groupby("PaymentMethod", observed=True)["Churn"].value_counts(normalize=True) * 100
```
